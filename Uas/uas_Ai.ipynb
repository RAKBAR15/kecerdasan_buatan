```python
# Class definition for Fuzzy logic
class Fuzzy:
    def __init__(self):
        self.min = 0
        self.max = 0

    def down(self, x):
        return (self.max - x) / (self.max - self.min)

    def up(self, x):
        return (x - self.min) / (self.max - self.min)

    def rev_down(self, value):
        return self.max - value * (self.max - self.min)

    def rev_up(self, value):
        return value * (self.max - self.min) + self.min
```

```python
# Define specific Fuzzy logic implementations
class Income(Fuzzy):
    def __init__(self):
        self.min = 0
        self.max = 10000

    def low(self, x):
        if x < self.min:
            return 1
        if x > self.max:
            return 0
        return self.down(x)

    def high(self, x):
        if x < self.min:
            return 0
        if x > self.max:
            return 1
        return self.up(x)

class Loan(Fuzzy):
    def __init__(self):
        self.min = 0
        self.max = 5000

    def low(self, x):
        if x < self.min:
            return 1
        if x > self.max:
            return 0
        return self.down(x)

    def high(self, x):
        if x < self.min:
            return 0
        if x > self.max:
            return 1
        return self.up(x)

class CreditHistory(Fuzzy):
    def __init__(self):
        self.min = 0
        self.max = 10

    def bad(self, x):
        if x < self.min:
            return 1
        if x > self.max:
            return 0
        return self.down(x)

    def good(self, x):
        if x < self.min:
            return 0
        if x > self.max:
            return 1
        return self.up(x)

class Eligibility(Fuzzy):
    def __init__(self):
        self.min = 0
        self.max = 100

    def low(self, fuzzy_value):
        return self.rev_down(fuzzy_value)

    def high(self, fuzzy_value):
        return self.rev_up(fuzzy_value)
```

```python
# Define rules for fuzzy logic
rule_map = {
    ("low", "high", "bad"): "low",
    ("low", "low", "bad"): "low",
    ("high", "high", "bad"): "low",
    ("high", "low", "bad"): "low",
    ("low", "high", "good"): "low",
    ("low", "low", "good"): "low",
    ("high", "high", "good"): "high",
    ("high", "low", "good"): "high",
}
```

```python
# Input values
income_value = 6000
loan_value = 2000
credit_history_value = 8
```

```python
# Apply rules and calculate result
fuzzy_real = 0
fuzzy = 0

for k, v in rule_map.items():
    income = Income()
    loan = Loan()
    credit_history = CreditHistory()
    eligibility = Eligibility()

    fuzzy_income = income.low(income_value) if k[0] == "low" else income.high(income_value)
    fuzzy_loan = loan.low(loan_value) if k[1] == "low" else loan.high(loan_value)
    fuzzy_credit = credit_history.bad(credit_history_value) if k[2] == "bad" else credit_history.good(credit_history_value)

    fuzzy_eligibility = min(fuzzy_income, fuzzy_loan, fuzzy_credit)
    eligibility_score = eligibility.low(fuzzy_eligibility) if v == "low" else eligibility.high(fuzzy_eligibility)

    fuzzy_real += fuzzy_eligibility * eligibility_score
    fuzzy += fuzzy_eligibility

    # Final result
result = int(fuzzy_real / fuzzy)
print(f"Kelayakan kredit adalah {result} dari 100.")
```
